```python
# Extracting the numeric value
data['term'] = data['term'].str.extract('(\d+)').astype(int)

```

```python
# Display the updated DataFrame
data.head()
```

```python
data['emp_length'].unique()
```

```python
# Extracting numeric values and handling '< 1 year'
data['emp_length'] = data['emp_length'].str.extract(r'(\d+)').astype(float)

# Replace NaN values (from '< 1 year') with 1
data['emp_length'] = data['emp_length'].fillna(1).astype(int)
```

```python
# Splitting issue_d into month and year
data[['issue_month', 'issue_year']] = data['issue_d'].str.split('-', expand=True)
```

```python
# Splitting earliest_cr_line into month and year
data[['cr_line_month', 'cr_line_year']] = data['earliest_cr_line'].str.split('-', expand=True)
```

```python
data.head()
```

```python
# Dropping the original issue_d column
data.drop(columns=['issue_d', 'earliest_cr_line', 'title'], inplace=True)
```
